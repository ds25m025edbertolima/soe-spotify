# Data Dictionary: Spotify Dataset

## Table: tracks

This is the main table. It contains one row per track/song and includes basic track, album, artist, genre, and audio feature information.

| Column                | Data Type | Description                                                                                                     |
| --------------------- | --------: | --------------------------------------------------------------------------------------------------------------- |
| `track_id`            |      text | Unique identifier for each Spotify track. Primary key for the track.                                            |
| `track_name`          |      text | Name/title of the track.                                                                                        |
| `track_popularity`    |      int4 | Spotify popularity score for the track. Usually ranges from 0 to 100, where higher means more popular.          |
| `duration_ms`         |      int4 | Duration of the track in milliseconds.                                                                          |
| `preview_url`         |      text | URL for a short preview clip of the track, if available.                                                        |
| `uri`                 |      text | Spotify URI for the track.                                                                                      |
| `album_id`            |      text | Identifier of the album the track belongs to. Foreign key to `albums.album_id`.                                 |
| `album_name`          |      text | Name of the album.                                                                                              |
| `album_type`          |      text | Type of album, such as album, single, or compilation.                                                           |
| `release_date`        |      text | Release date of the album or track. Stored as text.                                                             |
| `artist_id`           |      text | Identifier of the artist. Foreign key to `artists.artist_id`.                                                   |
| `artist_name`         |      text | Name of the artist.                                                                                             |
| `artist_popularity`   |      int4 | Spotify popularity score for the artist. Usually ranges from 0 to 100.                                          |
| `artist_followers`    |      int4 | Number of Spotify followers the artist has.                                                                     |
| `genres`              |      text | Genre or genre list associated with the artist or track.                                                        |
| `acousticness`        |    float8 | Audio feature measuring how acoustic the track is. Values closer to 1 mean more acoustic.                       |
| `danceability`        |    float8 | Measures how suitable a track is for dancing. Higher values mean more danceable.                                |
| `energy`              |    float8 | Measures intensity and activity of the track. Higher values usually mean louder, faster, or more energetic.     |
| `instrumentalness`    |    float8 | Predicts whether a track has no vocals. Values closer to 1 suggest the track is more instrumental.              |
| `key`                 |      int4 | Estimated musical key of the track. Usually represented as integers 0–11.                                       |
| `liveness`            |    float8 | Detects the presence of an audience. Higher values suggest the track may be live.                               |
| `loudness`            |    float8 | Overall loudness of the track, usually measured in decibels.                                                    |
| `mode`                |      int4 | Indicates modality of the track. Usually `1` means major and `0` means minor.                                   |
| `speechiness`         |    float8 | Measures the presence of spoken words in the track. Higher values mean more speech-like content.                |
| `tempo`               |    float8 | Estimated tempo of the track in beats per minute.                                                               |
| `time_signature`      |      int4 | Estimated time signature of the track.                                                                          |
| `valence`             |    float8 | Measures musical positivity. Higher values sound happier or more cheerful; lower values sound sadder or darker. |
| `mean_syllables_word` |    float8 | Average number of syllables per word in the lyrics.                                                             |
| `mean_words_sentence` |    float8 | Average number of words per sentence in the lyrics.                                                             |
| `n_sentences`         |    float8 | Number of sentences in the lyrics.                                                                              |
| `n_words`             |    float8 | Number of words in the lyrics.                                                                                  |
| `sentence_similarity` |    float8 | Measure of how similar or repetitive the sentences in the lyrics are.                                           |
| `vocabulary_wealth`   |    float8 | Measure of vocabulary richness or lexical diversity in the lyrics.                                              |


## Table: track_features

This table contains detailed audio and lyrical features for each track.

| Column                | Data Type | Description                                                |
| --------------------- | --------: | ---------------------------------------------------------- |
| `track_id`            |      text | Unique track identifier. Foreign key to `tracks.track_id`. |
| `acousticness`        |    float8 | Measures how acoustic the track sounds.                    |
| `danceability`        |    float8 | Measures how suitable the track is for dancing.            |
| `energy`              |    float8 | Measures intensity and activity of the track.              |
| `instrumentalness`    |    float8 | Predicts whether the track has no vocals.                  |
| `key`                 |      int4 | Estimated musical key of the track.                        |
| `liveness`            |    float8 | Measures likelihood that the track was performed live.     |
| `loudness`            |    float8 | Overall loudness of the track.                             |
| `mode`                |      int4 | Indicates major or minor modality.                         |
| `speechiness`         |    float8 | Measures the amount of spoken words in the track.          |
| `tempo`               |    float8 | Tempo in beats per minute.                                 |
| `time_signature`      |      int4 | Estimated time signature.                                  |
| `valence`             |    float8 | Measures how positive or cheerful the track sounds.        |
| `mean_syllables_word` |    float8 | Average number of syllables per lyric word.                |
| `mean_words_sentence` |    float8 | Average number of words per lyric sentence.                |
| `n_sentences`         |    float8 | Total number of lyric sentences.                           |
| `n_words`             |    float8 | Total number of lyric words.                               |
| `sentence_similarity` |    float8 | Measures sentence similarity or repetitiveness in lyrics.  |
| `vocabulary_wealth`   |    float8 | Measures lyrical vocabulary richness.                      |


## Table: albums

This table contains album-level information.

| Column              | Data Type | Description                                              |
| ------------------- | --------: | -------------------------------------------------------- |
| `album_id`          |      text | Unique identifier for the album. Primary key for albums. |
| `album_name`        |      text | Name of the album.                                       |
| `album_type`        |      text | Type of album, such as album, single, or compilation.    |
| `release_date`      |      text | Album release date. Stored as text.                      |
| `total_tracks`      |      int4 | Number of tracks on the album.                           |
| `uri`               |      text | Spotify URI for the album.                               |
| `artist_id`         |      text | Identifier of the main artist associated with the album. |
| `artist_name`       |      text | Name of the main artist associated with the album.       |
| `artist_followers`  |      int4 | Number of Spotify followers for the artist.              |
| `artist_popularity` |      int4 | Spotify popularity score for the artist.                 |


## Table: artists

This table contains artist-level information.

| Column              | Data Type | Description                                                |
| ------------------- | --------: | ---------------------------------------------------------- |
| `artist_id`         |      text | Unique identifier for the artist. Primary key for artists. |
| `artist_name`       |      text | Name of the artist.                                        |
| `artist_popularity` |      int4 | Spotify popularity score for the artist.                   |
| `followers`         |      int4 | Number of Spotify followers the artist has.                |
| `type`              |      text | Entity type, likely `artist`.                              |
| `genres`            |      text | Genre or list of genres associated with the artist.        |


## Table: genres

This table stores genres connected to artists.

| Column        | Data Type | Description                                                   |
| ------------- | --------: | ------------------------------------------------------------- |
| `id`          |      int4 | Unique identifier for each genre record.                      |
| `artist_id`   |      text | Identifier of the artist. Foreign key to `artists.artist_id`. |
| `artist_name` |      text | Name of the artist.                                           |
| `genre`       |      text | Genre associated with the artist.                             |
